**ANALISIS KUALITAS TIDUR, OLAHRAGA DAN RISIKO GANGGUAN PANIK DENGAN DECISION TREE DAN CRISP-DM**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
# Load dataset
df = pd.read_csv('DataMentah.csv')

display(df)
```

```python
# SELEKSI FITUR

# Seleksi fitur yang relevan
selected_features = [
    'Family History',
    'Personal History',
    'Current Stressors',
    'Symptoms',
    'Severity',
    'Impact on Life',
    'Psychiatric History',
    'Lifestyle Factors',
    'Panic Disorder Diagnosis'
]

# Memastikan bahwa kolom-kolom tersebut ada dalam dataset
df= df[selected_features]

# Menampilkan seluruh data
print("Seteleh seleksi Fitur : ")
display(df)
```

```python
# Cek Missing Values
missing_values_fill = df.isnull().sum()

missing_values_fill
```

```python
# Mengganti missing values dengan nilai 'No' pada kolom Psychiatric History'
df['Psychiatric History'].fillna('No', inplace=True)

print("Data Setelah Penanganan Missing Values : ")
display (df)
```

```python
# Cek Missing Values lagi
missing_values_after_fill = df.isnull().sum()

missing_values_after_fill
```

```python
# Mengecek duplikasi pada dataset
duplicate_rows = df.duplicated().sum()

# Menampilkan keterangan mengenai jumlah duplikat
duplicate_rows_message = f"Jumlah baris yang duplikat: {duplicate_rows}"

duplicate_rows_message

```

```python
# Hapus duplikasi
df = df.drop_duplicates()
display(df)
```

```python
# Hapus data dengan Lifestyle Factors = Diet
df= df[df["Lifestyle Factors"] != "Diet"]
display (df)
```

```python
kolom_kategorikal = [
    "Family History", "Personal History", "Current Stressors", "Symptoms", "Severity",
    "Impact on Life", "Psychiatric History", "Lifestyle Factors"
]


for kolom in kolom_kategorikal:
    df[kolom] = df[kolom].astype('category')


```

```python


# Pisahkan fitur dan target
X = df.drop(columns=["Panic Disorder Diagnosis"])
y = df["Panic Disorder Diagnosis"]

# Pecah data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.6, random_state=42, stratify=y)

# Cek jumlah data
print(f"Training set: {len(X_train)} data")
print(f"Testing set: {len(X_test)} data")
```

```python
X_train = X_train.apply(lambda col: col.astype('category'))
X_test = X_test.apply(lambda col: col.astype('category'))
y_train = y_train.astype('category')
y_test = y_test.astype('category')
```

```python
X_train_encoded = X_train.apply(lambda col: col.cat.codes)
X_test_encoded = X_test.apply(lambda col: col.cat.codes)
y_train_encoded = y_train.cat.codes
y_test_encoded = y_test.cat.codes
```

```python
# Model Decision Tree
model = DecisionTreeClassifier()
model.fit(X_train_encoded, y_train_encoded)
```

```python
# Prediksi
y_pred = model.predict(X_test_encoded)
```

```python
from sklearn.metrics import confusion_matrix

# Menghitung confusion matrix
cm = confusion_matrix(y_test_encoded, y_pred)

# Menampilkan hasil TP, TN, FP, FN
tn, fp, fn, tp = cm.ravel()

print(f"True Positive (TP): {tp}")
print(f"True Negative (TN): {tn}")
print(f"False Positive (FP): {fp}")
print(f"False Negative (FN): {fn}")
```

```python
# Evaluasi Model
accuracy = accuracy_score(y_test_encoded, y_pred)
precision = precision_score(y_test_encoded, y_pred)
recall = recall_score(y_test_encoded, y_pred)
f1 = f1_score(y_test_encoded, y_pred)

# Menampilkan hasil
print(f'Akurasi: {accuracy:.2%}')
print(f'Precision: {precision:.2%}')
print(f'Recall: {recall:.2%}')
print(f'F1-Score: {f1:.2%}')
```
